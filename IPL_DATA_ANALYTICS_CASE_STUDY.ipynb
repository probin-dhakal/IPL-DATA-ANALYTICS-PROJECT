**Most of the runs scored by Rohit Sharma come from singles and boundaries (fours and sixes), indicating a batting style that balances effective strike rotation with aggressive power hitting. Contributions from doubles and triples are comparatively minimal, suggesting a stronger reliance on boundary scoring and consistent accumulation through singles.**

## Section 2 — Does Winning the Toss Imply Winning the Match?







```python
match_data.columns
```

```python
match_data[['team1', 'team2' , 'toss_winner' , 'winner']]
```

```python
## if "toss_winner" equals to "match_winner" , then assign "Yes" else assign "No"

match_data['toss_win_game_win'] = np.where(match_data['toss_winner']==match_data['winner'] , 'Yes' , 'No')
```

```python
match_data.columns
```

```python
match_data['toss_win_game_win'].value_counts()
```

## lets create Pie chart of above values ..

```python
match_data['toss_win_game_win'].value_counts().index
```

```python
match_data['toss_win_game_win'].value_counts().values
```

```python
labels = match_data['toss_win_game_win'].value_counts().index
values = match_data['toss_win_game_win'].value_counts().values

trace = go.Pie(labels=labels, values=values)

data = [trace]

fig = go.Figure(data=data)

fig.update_traces(
    labels=['Toss Winner Won Match', 'Toss Winner Lost Match'],
    hoverinfo='label+percent',
    textinfo='label+percent'
)

fig.show()
```

# CONCLUSION :
Winning the toss may offer an initial advantage, but it does not strongly determine the final match result. Overall team performance plays a much larger role in securing victory.

# Section 3 — Team Win Statistics & Win Percentages

```python
match_data.columns
```

```python
match_data.head(5)
```

```python
match_data[['team1', 'team2']]
```

```python
matches_played = match_data['team1'].value_counts() + match_data['team2'].value_counts()
```

```python
type(matches_played)
```

```python
matches_played_df = matches_played.to_frame().reset_index()
```

```python
matches_played_df.columns = ['team_name' , 'Matches_played']
```

```python
matches_played_df
```

```python
wins = pd.DataFrame(match_data['winner'].value_counts()).reset_index()
```

```python
wins.columns = ['team_name' , 'Wins']
```

```python
wins
```

```python
played =matches_played_df.merge(wins, on='team_name' , how='inner' )
```

```python
played
```

```python
import plotly.graph_objects as go

# Calculate win %
played['Win_Percentage'] = (
    played['Wins'] / played['Matches_played'] * 100
).round(2)

# Team with highest win %
best_team = played.loc[played['Win_Percentage'].idxmax()]

fig = go.Figure()

fig.add_trace(go.Bar(
    y=played['team_name'],
    x=played['Matches_played'],
    name='Matches Played',
    orientation='h'
))

fig.add_trace(go.Bar(
    y=played['team_name'],
    x=played['Wins'],
    name='Wins',
    orientation='h',
    text=played['Win_Percentage'].astype(str) + '%',
    textposition='outside'
))

fig.update_layout(
    title=(
        f'IPL Teams: Matches Played vs Wins<br>'
        f'<sup>Highest Win %: {best_team["team_name"]} '
        f'({best_team["Win_Percentage"]}%)</sup>'
    ),
    barmode='group',
    xaxis_title='Count',
    yaxis_title='Teams',
    height=700
)

fig.show()

print(
    f"Highest Win Percentage: {best_team['team_name']} "
    f"({best_team['Win_Percentage']}%)"
)
```
